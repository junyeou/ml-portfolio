```python
import numpy as np

sqft = np.array([850, 1200, 1500, 2000, 2400])
prices = np.array([200000, 280000, 330000, 420000, 500000])

price_per_sqft = prices / sqft

houses = np.array ([
    [850, 2, 200000],
    [1200, 3, 280000],
    [1500, 3, 330000],
])

print (houses.shape)
print(houses[0])     #first row
print(houses[:, 0])  #first column

print(price_per_sqft)

print(sqft)
print(sqft.shape)      # how many items, and in what layout
print(sqft.mean())     # average
print(sqft * 2)        # math applies to EVERY element at once
print(sqft[sqft > 1300])  # filtering with a condition
```

```python
print(houses[1, 2])
print(houses[2, 0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram: how are prices distributed?
plt.hist(df["price"], bins=5)
plt.xlabel("Price")
plt.ylabel("Number of houses")
plt.title("Distribution of house prices")
plt.show()

# Scatter plot: relationship between two numeric columns
plt.scatter(df["sqft"], df["price"])
plt.xlabel("Square feet")
plt.ylabel("Price")
plt.title("Size vs. price")
plt.show()
```

```python
df.corr()
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation heatmap")
plt.show()
```
